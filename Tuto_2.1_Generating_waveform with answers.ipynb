```python
from pycbc.waveform import get_td_waveform
# The outputs of this function are the "plus" and "cross" polarizations 
# of the gravitational-wave signal as viewed from the line of sight at 
# a given source inclination
# (assumed inclination of pi/3 here, the default is zero inclination if none is provided)
hp, hc = get_td_waveform(approximant="IMRPhenomT",
    mass1=10,
    mass2=15,
    inclination=np.pi/3,
    delta_t=1.0/16348,
    f_lower = 30)
plt.figure(figsize=plt.figaspect(0.4))
plt.plot(hp.sample_times,hp,label="plus polarisation")
plt.plot(hp.sample_times,hc,label="cross polarisation")
plt.xlabel('Time(s)')
plt.ylabel('Strain')
plt.legend()
plt.grid()


#lets zoom at the merger

plt.figure(figsize=plt.figaspect(0.4))
plt.plot(hp.sample_times,hp,label="plus polarisation")
plt.plot(hp.sample_times,hc,label="cross polarisation")
plt.xlabel('Time(s)')
plt.ylabel('Strain')
plt.xlim(-0.1,0.1)
plt.legend()
plt.grid()
```

```python
from pycbc.waveform import get_fd_waveform

hp, hc = get_fd_waveform(approximant = 'IMRPhenomXAS',
                         mass1 = 15,
                         mass2 = 15,
                         f_lower = 30,
                         inclination = np.pi/3,
                         delta_f = 1./8,)

plt.figure(figsize = plt.figaspect(0.4))
#plot the norm of the Fourier components of the signal polarizations.
plt.loglog(hp.sample_frequencies, hp.squared_norm(), label = 'plus polarisation')
plt.loglog(hp.sample_frequencies, hc.squared_norm(), label = 'cross polarisation')
plt.xlabel("Frequency [Hz]")
plt.ylabel(r'$|\tilde{h}(f)|^2$ [1/Hz]')
plt.xlim(30,2048)
plt.legend()
plt.grid()
```

```python
from pycbc.waveform import get_fd_waveform

hp, hc = get_fd_waveform(approximant="IMRPhenomXAS",
                         mass1=15,
                         mass2=15,
                         f_lower=30,
                         inclination=np.pi/3,
                         delta_f=1./8)

plt.figure(figsize=plt.figaspect(0.4))

# We will plot the norm of the Fourier components of the signal polarizations.
plt.loglog(hp.sample_frequencies, hp.squared_norm(), label='Plus Polarization')
plt.loglog(hp.sample_frequencies, hc.squared_norm(), label='Cross Polarization')
plt.xlabel('Frequency [Hz]')
plt.ylabel(r'$|\tilde{h}(f)|^2$ [1/Hz]')
plt.xlim([30,2048])
plt.legend()
plt.grid()
```

**Question 1**.
Generate and plot the waveform associated with the binary neutron star merger GW170817.

How much time does the signal last using a lower frequency of 30 Hz?


```python
hp, hc = get_td_waveform(approximant="IMRPhenomT",
    mass1=1.46,
    mass2=1.27,
    inclination=np.pi/3,
    delta_t=1.0/16348,
    f_lower = 30)
plt.figure(figsize=plt.figaspect(0.4))
plt.plot(hp.sample_times,hp,label="plus polarisation")
plt.plot(hp.sample_times,hc,label="cross polarisation")
plt.xlabel('Time(s)')
plt.ylabel('Strain')
plt.legend()
plt.grid()


#lets zoom at the merger

plt.figure(figsize=plt.figaspect(0.4))
plt.plot(hp.sample_times,hp,label="plus polarisation")
plt.plot(hp.sample_times,hc,label="cross polarisation")
plt.xlabel('Time(s)')
plt.ylabel('Strain')
plt.xlim(-0.1,0.1)
plt.legend()
plt.grid()
print(hp.duration)
print(hc.duration)
```
